### 0. Imports

```python
from typing import List, Dict
```

### 1. Definición de los datos

```python
# Clase que define los atributos de los casos
class Case:
    def __init__(self, name: str, ingredients: List[str], steps: List[str], outcome: str):
        self.name = name
        self.ingredients = ingredients
        self.steps = steps
        self.outcome = outcome

    def __repr__(self):
        return f"Case ({self.name})"
```

```python
case_base = [
    Case(
        name="Plain Pancakes",
        ingredients=["flour", "milk", "egg", "baking powder", "salt", "butter"],
        steps=[
            "Mix dry ingredients",
            "Add milk and egg",
            "Whisk until smooth",
            "Cook batter on pan",
        ],
        outcome="Golden fluffy plain pancakes"
    ),
    Case(
        name="Omelette",
        ingredients=["egg", "salt", "butter", "cheese"],
        steps=[
            "Crack eggs into a bowl",
            "Add salt and whisk",
            "Melt butter on pan",
            "Pour egg mixture and cook",
            "Add cheese and fold"
        ],
        outcome="Delicious cheesy omelette"
    ),
    Case(
        name="French Toast",
        ingredients=["bread", "egg", "milk", "cinnamon", "butter"],
        steps=[
            "Mix egg, milk, and cinnamon",
            "Dip bread slices into mixture",
            "Cook on buttered pan until golden"
        ],
        outcome="Sweet and crispy french toast"
    )
]
```

### 2. Definición de problema/caso nuevo

```python
target_problem = {
    "goal": "Make blueberry pancakes",
    "new_ingredient": "blueberries"
}
```

### 3. Proceso de las 4R

#### 3.1. Retrieve
 
Buscamos en la base de datos el caso más parecido al problema actual.

```python
# Función que calcula la similitud entre casos
def retrieve_most_similar(case_base: List[Case], new_ingredient: str) -> Case:
    max_overlap = 0
    best_case = None
    # Miramos todos los casos
    for case in case_base:
        # Calculamos la similitud como el numero de ingredientes en comun (calculando la cantidad de ingredientes que se solapan)
        overlap = len(set(case.ingredients) & set(["flour", "milk", "egg", new_ingredient]))
        
        # Mostramos paso
        print(f"Receta: {case.name}\n" + 
              f"  Ingredientes: {set(case.ingredients)}\n" +
              f"  Nuevos Ingredientes: {set(['flour', 'milk', 'egg', new_ingredient])}\n" +
              f"  Ingredientes en comun: {set(case.ingredients) & set(['flour', 'milk', 'egg', new_ingredient])}, " + 
              f"se solapa{'n' if overlap > 1 else ''} {overlap} ingrediente{'s' if overlap > 1 else ''}\n")

        # Miramos si es el mejor hasta ahora
        if overlap > max_overlap:
            max_overlap = overlap
            best_case = case

    return best_case
```

```python
retrieved_case = retrieve_most_similar(case_base, target_problem["new_ingredient"])
```

#### 3.2. Reuse (+ adaptación)
 
Una vez se ha recuperado el caso más parecido, hacemos las modificaciones (añadir ingresiente y paso nuevo)

```python
adapted_ingredients = retrieved_case.ingredients + [target_problem["new_ingredient"]]
adapted_steps = retrieved_case.steps.copy()
adapted_steps.insert(3, "Add blueberries to the batter")

# Mostramos adaptaciones del caso que recuperamos
print("Ingredients:", adapted_ingredients)
print("Steps:", adapted_steps)
```

#### 3.3. Revise

Adaptamos los pasos

```python
print("\n--- Revision ---")
print("Problem: Batter turns blue. Revising step to avoid discoloration...")
adapted_steps[3] = "Pour batter on pan, then sprinkle blueberries on top"

# Final outcome
outcome = "Tasty blueberry pancakes with intact batter color"
```

#### 3.4. Retain

Guardamos caso nuevo

```python
new_case = Case(
    name="Blueberry Pancakes",
    ingredients=adapted_ingredients,
    steps=adapted_steps,
    outcome=outcome
)
case_base.append(new_case)
```

### 4. Result

```python
print(f"Name: {new_case.name}\n" +
      f"Ingredients: {new_case.ingredients}\n" + 
      f"Steps:")

for i, step in enumerate(new_case.steps, 1):
    print(f"  {i}. {step}")

print("Outcome:", new_case.outcome)
```
