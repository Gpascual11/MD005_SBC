```python
from fpdf import FPDF

class PDF(FPDF):
    def header(self):
        # Título principal estilo informe
        self.set_font('Helvetica', 'B', 16)
        self.set_text_color(44, 62, 80)  # Azul oscuro
        self.cell(0, 10, 'Apuntes: Sistemas Basados en el Conocimiento', 0, 1, 'C')
        self.ln(10)

    def footer(self):
        self.set_y(-15)
        self.set_font('Helvetica', 'I', 8)
        self.set_text_color(128, 128, 128)
        self.cell(0, 10, f'Pagina {self.page_no()}', 0, 0, 'C')

    def chapter_title(self, label):
        # Fondo gris claro y letra azul para separar secciones
        self.set_font('Helvetica', 'B', 12)
        self.set_fill_color(234, 237, 237) # Gris fondo
        self.set_text_color(31, 97, 141)   # Azul medio
        self.cell(0, 8, label, 0, 1, 'L', fill=True)
        self.ln(4)

    def chapter_body(self, text):
        self.set_font('Helvetica', '', 10)
        self.set_text_color(50, 50, 50) # Gris oscuro texto
        # Reemplazar caracteres problemáticos básicos
        text = text.replace('”', '"').replace('“', '"')
        self.multi_cell(0, 5, text)
        self.ln()

    def print_formula_box(self, formula):
        # Caja gris para formulas
        self.set_fill_color(244, 246, 246) # Gris muy claro
        self.set_text_color(0, 0, 0)
        self.set_font('Courier', 'B', 10) # Fuente tipo código
        self.ln(2)
        self.cell(0, 8, formula, 0, 1, 'C', fill=True)
        self.ln(2)

# Crear PDF
pdf = PDF()
pdf.add_page()

# --- CONTENIDO ---

# 1. INTRO
pdf.chapter_title('1. Introduccion: Tipos de Aprendizaje y Problemas')
pdf.chapter_body("""El objetivo principal es etiquetar datos de entrada (features) en categorias discretas (clases), a diferencia de la regresion que predice valores continuos.

Tipos de Aprendizaje:
- No Supervisado: No hay etiquetas (ground truth). Buscamos patrones.
- Supervisado: Entrenamos con datos que ya tienen la etiqueta correcta.

Tipos de Clasificacion:
- Binaria: Dos clases (0/1). Ej: Regresion Logistica.
- Multiclase: >2 grupos, cada dato pertenece a una sola clase. Ej: Arboles, KNN, SVM.
- Multilabel: Un dato puede pertenecer a varias clases a la vez (Ej. Segmentacion).""")

# 2. REGRESION LOGISTICA
pdf.chapter_title('2. Regresion Logistica')
pdf.chapter_body("""Modelo base para clasificacion binaria. Se usa para clasificar, no para regresion.

Conceptos Clave:
- Funcion Sigmoide: Transforma valores a una probabilidad (0 a 1).""")

pdf.print_formula_box("P(y=1|X) = 1 / (1 + e^-z)")

pdf.chapter_body("""- Threshold (Umbral): Limite para decidir la clase.
   * Si P > 0.5 -> Clase 1
   * Si P < 0.5 -> Clase 0
   Nota: En problemas desbalanceados, mover el threshold es vital.

- Funcion de Coste (Log Loss): Se minimiza iterativamente comparando con el ground truth.

Interpretacion de Coeficientes:
- B0 (Intercepto): Punto de partida.
- Bp (Coeficientes): Cuanto cambia la probabilidad al aumentar una unidad del predictor.
   (+) Aumenta prob. clase 1.
   (-) Disminuye prob. clase 1.

Pros: Interpretable, probabilistico, eficiente.
Contras: Malo si no hay separacion lineal.""")

# 3. METRICAS
pdf.chapter_title('3. Metricas y Estrategia: El caso del Cancer')
pdf.chapter_body("Recall (Sensibilidad): De los casos reales positivos, cuantos detectamos?")

pdf.print_formula_box("Recall = TP / (TP + FN)")

pdf.chapter_body("""Importancia: En medicina, un Falso Negativo es fatal. Priorizamos Recall sobre Precision.

Estrategias ante errores:
1. Analizar el Threshold: Bajarlo para capturar mas positivos (ej. > 0.3).
2. Data Augmentation: Crear datos sinteticos para equilibrar clases.""")

# 4. SVM
pdf.chapter_title('4. Maquina de Soporte Vectorial (SVM)')
pdf.chapter_body("""Busca el mejor limite de separacion (hiperplano).

Conceptos:
- Vectores de Soporte: Puntos mas cercanos al hiperplano. Son los unicos que definen el modelo.
- Margen: Distancia entre hiperplano y vectores de soporte (buscamos maximizarlo).

Hard Margin vs Soft Margin:
- Hard: Asume separacion perfecta (Overfitting).
- Soft: Permite cierto error para generalizar mejor (parametro C).

Kernel Trick:
Cuando los datos no son separables linealmente, proyecta los datos a una dimension superior.
Tipos: Lineal, Polinomico, RBF (Gaussiano, el mas usado), Sigmoidal.""")

# 5. MULTICLASE
pdf.chapter_title('5. Clasificacion Multiclase y No Lineal')
pdf.chapter_body("""A. Arboles de Decision:
- Divide el espacio basandose en pureza (Gini/Entropia).
- Problema: Overfitting. Solucion: Prunning (poda).

B. K-Nearest Neighbors (KNN):
- Clasifica un punto segun la mayoria de sus K vecinos mas cercanos.

Distancias:
- Euclidiana (Recta), Manhattan (Callejeando).
- Coseno: Mide el angulo. Vital en NLP y LLMs (Similitud semantica).

Pros y Contras KNN:
(+) No asume distribucion.
(-) Computacionalmente caro.
(-) Sensible a outliers y dimension.""")

# Guardar
pdf.output("Apuntes_SBC_Bonito.pdf")
print("PDF generado correctamente: Apuntes_SBC_Bonito.pdf")
```
